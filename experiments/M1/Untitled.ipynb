```python
import numpy as np

def make_points(n, low=-2.0, high=2.0, seed=None):
    """Return an (n, 3) array of points uniform in [low, high] per dimension."""
    rng = np.random.default_rng(seed)
    return rng.uniform(low, high, size=(n, 3))

points = make_points(100)
print(points.shape)   # (100, 3)
```

```python
def to_latex(points, limit=None):
    return "\n".join(f"{x} {y} {z}" for x, y, z in points[:limit])
```

```python
from loka.geometrical_methods import herding_order, k_center_order
import torch
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)
```

```python
order = herding_order(points, max_ratio=1, device=device)
print(to_latex(points[order], 10))
```

```python
order = k_center_order(points, max_ratio=1, device=device)
print(to_latex(points[order], 10))
```
