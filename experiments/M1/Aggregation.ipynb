```python
import os
import pandas as pd
import matplotlib.pyplot as plt

data_dir = "/home/axelbjarkar/RU/T-404-LOKA-code/data/results"
files = [
    'random_subset_50epoch.csv',
    'herding_subset_50epoch.csv',
    'k_center_no_control_points_subset_50epochs.csv',
    'k_center_rand_control_points_subset_50epochs.csv'
]

fig, ax = plt.subplots(figsize=(16, 9))

for f in files:
    df = pd.read_csv(os.path.join(data_dir, f))
    label = os.path.splitext(f)[0]  # file name without .csv
    ax.plot(df['ratio'], df['mse'], 'o-', linewidth=2, markersize=8,
            markerfacecolor='white', markeredgewidth=2, label=label)

ax.set_xlabel(r"$\frac{|S|}{|E|}$", fontsize=20)
ax.set_ylabel("MSE", fontsize=20)
ax.set_title("Model Performance vs Training Subset Size\n(50 epochs, 130k samples)", fontsize=22)
ax.legend(fontsize=14)

ax.grid(True, alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
output_dir = '/home/axelbjarkar/RU/T-404-LOKA-code/experiments/M1/results'
plt.savefig(f"{output_dir}/aggregation_50epoch.png")
plt.show()
```
