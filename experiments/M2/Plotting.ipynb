```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from dotenv import load_dotenv

load_dotenv()
BASE_DIR = os.getenv("BASE_DIR")
data_dir = f"{BASE_DIR}/data/ordered-data/results/mse_all"

files = [
    'forgetting.csv',
    'herding.csv',
    'herding_with_counts_10p.csv',
    'herding_with_counts_and_near_goal_10p.csv',
    'herding_with_near_goal_10p.csv',
    'k_center_no_S0.csv',
    'k_center_no_S0_with_counts_10p.csv',
    'k_center_no_S0_with_counts_and_near_goal_10p.csv',
    'k_center_no_S0_with_near_goal_10p.csv',
    'k_center_rand_S0_25p.csv',
    'k_center_rand_S0_with_counts_10p.csv',
    'k_center_rand_S0_with_counts_and_near_goal_10p.csv',
    'k_center_rand_S0_with_near_goal_10p.csv',
    'random.csv'
]

data_dir2 = f'{BASE_DIR}/data/forgetting_counts/herding_and_forgetting'

files2 = [
    'herding_95_forgetting_5.csv'
]

fig, ax = plt.subplots(figsize=(16, 9))

for f in files:
    df = pd.read_csv(os.path.join(data_dir, f))
    label = os.path.splitext(f)[0]  # file name without .csv
    ax.plot(df['ratio'], df['mse'], 'o-', linewidth=2, markersize=8,
            markerfacecolor='white', markeredgewidth=2, label=label)

for f in files2:
    df = pd.read_csv(os.path.join(data_dir2, f))
    label = os.path.splitext(f)[0]  # file name without .csv
    ax.plot(df['ratio'], df['mse'], 'o-', linewidth=2, markersize=8,
            markerfacecolor='white', markeredgewidth=2, label=label)

ax.set_xlabel(r"$\frac{|S|}{|E|}$", fontsize=20)
ax.set_ylabel("MSE", fontsize=20)
ax.set_title("Model Performance vs Training Subset Size\n(50 epochs, 1.8M samples)", fontsize=22)
ax.legend(fontsize=14)

ax.grid(True, alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

#ax.set_yscale("log")

plt.show()
```
