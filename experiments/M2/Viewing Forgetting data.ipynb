```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
from loka.data import load_forgetting_states
from dotenv import load_dotenv
import os

load_dotenv()

BASE = os.getenv("BASE_DIR")
X, Y, C = load_forgetting_states(f"{BASE}/data/forgetting_counts/forgetting_e.txt")

c = C.cpu().numpy() if hasattr(C, "cpu") else np.asarray(C)
max_c = int(c.max())
bins = np.arange(max_c + 2) - 0.5          # one bin per integer count

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

for ax, log in zip(axes, [False, True]):
    ax.hist(c, bins=bins, color="tab:blue", linewidth=0)   # no edges: they clutter dense bins
    ax.set_xlabel("forgetting count")
    ax.set_ylabel("# examples")
    ax.xaxis.set_major_locator(MaxNLocator(nbins=10, integer=True))
    ax.set_xlim(-0.5, max_c + 0.5)
    if log:
        ax.set_yscale("log")
        ax.set_title("Log scale")
    else:
        ax.set_title("Linear scale")

fig.suptitle("Forgetting count distribution")
plt.tight_layout()
plt.show()
```
